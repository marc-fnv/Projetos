```python
dados = pd.read_csv('datasets/fifa19-utf8.csv')

# Cria a nova coluna antes de eliminar as outras
dados['Total'] = dados['Acceleration'] + dados['Agility'] + dados['Reactions']

# Mantém apenas as colunas que queremos visualizar
dados = dados[['Name', 'Total']]

# Ordena do maior para o menor e pega os 10 primeiros
dados = dados.sort_values('Total', ascending=False).head(10)

# dados
```
