```python
import numpy as np
import sys
```

```python
listaA = [1, 2, 3, 4, 5]
listaB = np.array([1, 2, 3, 4, 5])
```

```python
print(listaA) # mostra a lista com virgulas
print(listaB) # mostra a lista sem virgulas
```

```python
sys.getsizeof('Lista python', listaA) # mostra o tamanho da lista em bytes
```

```python
sys.getsizeof('Lista numpy', listaB) # mostra o tamanho da lista em bytes
```

```python
listaA = list(range(1_000_000))
listaB = np.arange(1_000_000)

print('Lista python', sys.getsizeof(listaA))
print('Lista numpy', listaB.nbytes)
```

```python
tamanho_lista = sys.getsizeof(listaA) + sum(sys.getsizeof(x) for x in listaA)

print('Lista python', tamanho_lista)
print('Lista numpy', listaB.nbytes)
```
