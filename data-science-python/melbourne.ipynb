```python
import pandas as pd
import re
```

```python
df = pd.read_csv('datasets/melbourne_house.csv')
```

```python
filter1 = df.loc[(df['Rooms'] == 3) & (df['Type'] == 'h') & (df['Price'] > 500000)] # filtra casas com 3 quartos, tipo 'h' e preço maior que 500000
filter1
```

```python
filter2 = df.loc[df['Address'].str.contains('Turner st|Turner Rd', flags=re.I)] # filtra casas com endereço contendo 'Turner'
filter2
```

```python
filter3 = df.loc[df['Address'].str.contains('^59', flags=re.I) & (df['Price'] < 500000)] # filtra casas com endereço contendo '59' e preço menor que 500000
filter3
```

```python
df.loc[df['SellerG'] == 'Nelson', 'SellerG'] = 'Ronald' # troca o vendedor 'Nelson' por 'Ronald' e retorna apenas a coluna 'SellerG'
df
```

```python
df.loc[df['SellerG'] == 'Ronald', ['Method', 'Type']] = 'Pending' # troca o método de venda para 'Pending' e 'Type' para todas as casas vendidas pelo vendedor 'Ronald'
df
```

```python
df.groupby(['SellerG']).mean(numeric_only=True) # agrupa os dados pelo vendedor e calcula a média dos preços das casas vendidas por cada vendedor
df
```

```python
df.groupby(['SellerG']).sum().sort_values('Price', ascending = False).head(10) # agrupa os dados pelo vendedor e calcula a soma dos preços das casas vendidas por cada vendedor, mostrando os 10 vendedores com maior soma de preços
df
```
