```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
gas = pd.read_csv('datasets/gas_prices.csv')
gas 
```

```python
plt.title('Preço do gás por país em dólar')
plt.plot(gas['Year'], gas['Australia'], 'r.--',label='Australia')
plt.plot(gas['Year'], gas['Italy'], 'y.--',label='Italy')
plt.plot(gas['Year'], gas['USA'], 'b.--', label='USA')
plt.plot(gas['Year'], gas['Japan'], 'g.--', label='Japan')
plt.xticks(gas['Year'][::2])
plt.xlabel('Ano')
plt.ylabel('US Dólar')
plt.legend()
plt.show()
```

```python
plt.title('Preço do gás por país em dólar')

list_country = ['Australia','USA', 'Italy', 'Mexico', 'France', 'Germany']

for country in list_country:
    plt.plot(gas['Year'], gas[country], label=country, marker='.')

plt.xticks(gas['Year'][::2])
plt.xlabel('Ano')
plt.ylabel('US Dólar')
plt.savefig('preco_gas.png')
plt.savefig('preco_gas.pdf')
plt.legend()
plt.show()
```
